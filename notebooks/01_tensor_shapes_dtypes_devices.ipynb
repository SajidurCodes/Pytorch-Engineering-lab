```python
import sys
import torch

print(sys.executable)
print(torch.__version__)
```

```python
features = torch.tensor(
    [[1, 2, 3],
     [4, 5, 6]],
    dtype=torch.float32,
    device="cpu",
)

print(tuple(features.shape))
print(features.dtype)
print(features.device)
```

```python
first_row = features[0]
first_column = features[:, 0]
first_row_slice = features[0:1]

print("features[0]   ", first_row.tolist(), tuple(first_row.shape))
print("features[:, 0]", first_column.tolist(), tuple(first_column.shape))
print("features[0:1] ", first_row_slice.tolist(), tuple(first_row_slice.shape))
```

### Dtype inference

```python
counts = torch.tensor([1, 2, 3])
measurements = torch.tensor([1.0, 2.0, 3.0])

print("counts:", counts.dtype)
print("measurements:", measurements.dtype)
```

```python
values = [0, 1, 2]

class_ids = torch.tensor(values, dtype=torch.long)
assert class_ids.dtype == torch.long
temperatures = torch.tensor(values, dtype=torch.float32)



print(class_ids, class_ids.dtype)
print(temperatures, temperatures.dtype)
print(temperatures.mean())
```

### Device placement

```python
labels = torch.tensor([0, 1], dtype=torch.long, device=features.device)

print(tuple(labels.shape), labels.dtype, labels.device)
assert labels.shape[0] == features.shape[0]
assert labels.device == features.device
```

### Random seeds

```python
torch.manual_seed(42)
first = torch.rand(3)
second = torch.rand(3)

torch.manual_seed(42)
replay = torch.rand(3)

print("first equals second:", torch.equal(first, second))
print("first equals replay:", torch.equal(first, replay))
print(first)
```

Consecutive draws advance the generator. Resetting it to the same seed reproduces the first draw in this CPU experiment.
