```python
import torch

features = torch.tensor([[10, 20, 30],
                         [40, 50, 60]])

first_row = features[0:1]
first_row[0, 1] = 99

print("first_row:", first_row.tolist())
print("features: ", features.tolist())
```

```python
features[0, 0] = -1
print(first_row.tolist())
```

```python
copied_row = first_row.clone()
copied_row[0, 2] = 77

print("Copy:", copied_row.tolist())
print("View:", first_row.tolist())
print("Original:", features.tolist())
```

```python
print(features[0, 2].item())
```

```python
assert copied_row[0, 2].item() == 77
assert first_row[0, 2].item() == 30
assert features[0, 2].item() == 30
```

```python
scores = torch.tensor([[10, 20, 30],
                       [40, 50, 60]])

selected = scores[[0]]
selected[0, 0] = 99

print("Selected:", selected.tolist())
print("Original:", scores.tolist())
```

```python
second_row = scores[1:2]
second_row[0, 0] = 88
print(scores[1, 0].item())
```

```python
grid = torch.tensor([[10, 20, 30],
                     [40, 50, 60]])

print("Shape:", tuple(grid.shape))
print("Stride:", grid.stride()) 
# Moving right one column, from 10 to 20, how many positions do you advance? (row step, column step)
```

```python
transposed = grid.T

print("Values:", transposed.tolist())
print("Shape:", tuple(transposed.shape))
print("Stride:", transposed.stride())
```
